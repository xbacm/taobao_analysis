## 🛍️ 项目阶段三：用户行为路径转移矩阵（全局宏观版）

### 1. 数据口径
基于 `user_behavior_clean` 表（全量数据），使用 `LAG()` 窗口函数按 `user_id` 排序，提取所有用户的连续行为对，统计独立用户数（UV）。

### 2. 核心路径解读（Top 5 流量）
| 路径 (上一步 -> 当前步) | 独立用户数 (UV) | 业务洞察 |
| :--- | :--- | :--- |
| **pv -> pv** | 9,206 | 用户大量浏览、比价，流量在平台内反复流转。 |
| **cart -> pv** | 5,677 | **痛点！** 加购后大量用户返回浏览，购物车沦为“暂存区”，流失严重。 |
| **pv -> buy** | 4,515 | **亮点！** 直接购买人数极高，说明冲动消费属性极强，无需加购犹豫。 |
| **pv -> cart** | 4,015 | 加购转化率不错，但后续购买流失（说明用户愿意加，但不愿意立刻买）。 |
| **buy -> pv** | 3,959 | 购买后继续浏览，可能是凑单、比价，或是买完顺便逛逛。 |

### 3. 业务结论与建议
- **购物车环节需要重点优化**：`cart -> buy` 只有 1,139 人，而 `cart -> pv` 高达 5,677 人。建议运营对“加购未购买”的用户发放定向优惠券，或设置“购物车库存紧张提醒”，缩短用户的犹豫周期。
- **立刻购买路径不可忽视**：`pv -> buy` 的用户有 4,515 人，建议产品经理优化“立即购买”的结算流程，甚至针对这部分冲动型消费者设置“一键支付”或“快捷支付”入口。
