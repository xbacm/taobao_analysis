## 🛍️ 项目阶段三：用户行为路径转移矩阵（全局宏观版）

### 1. 数据口径
基于 `user_behavior_clean` 表（全量数据），使用 `LAG()` 窗口函数按 `user_id` 排序，提取所有用户的连续行为对，统计独立用户数（UV）。

### 2. 核心路径解读（Top 5 流量）
| 路径 (上一步 -> 当前步) | 独立用户数 (UV) | 业务洞察 |
| :--- | :--- | :--- |
| **pv -> pv** | 9,206 | 用户针对同一商品存在多次浏览行为，说明商品决策过程中存在重复查看和比较行为。|
| **cart -> pv** | 5,677 | **痛点！** 加购后继续浏览行为较多，说明用户在加入购物车后仍处于比较和决策阶段，购物车可能承担了临时收藏和比较工具的作用。 |
| **pv -> buy** | 4,515 | **亮点！** 存在较大比例的浏览后直接购买行为，说明部分商品决策链路较短，用户无需经过加购阶段即可完成购买。 |
| **pv -> cart** | 4,015 | 加购转化率不错，但后续购买流失（说明用户愿意加，但不愿意立刻买）。 |
| **buy -> pv** | 3,959 | 用户购买完成后仍继续浏览其他商品，可能存在关联购买、凑单或后续消费需求。 |

### 3. 业务结论与建议
- **购物车环节需要重点优化**：`cart -> buy` 只有 1,139 人，而 `cart -> pv` 高达 5,677 人。针对加购后未购买用户，可进一步分析停留时间和后续行为，并考虑优惠券、提醒机制等召回策略。
- **立刻购买路径不可忽视**：`pv -> buy` 的用户有 4,515 人，建议产品经理优化“立即购买”的结算流程，甚至针对这部分冲动型消费者设置“一键支付”或“快捷支付”入口。
