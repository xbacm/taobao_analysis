## 🚀 项目阶段四：生存分析与返回分析

### 1. 生存分析（用户生命周期）
- **分析逻辑**：计算每个用户最后一次活跃与第一次活跃之间的天数差（生命周期）。
- **数据发现**：生命周期为 8 天的用户高达 6,944 人，呈上升趋势。
- **业务洞察**：由于数据集仅有 9 天跨度，用户生命周期普遍偏长（绝大部分从首日留存到了末日）。这说明样本中的淘宝用户粘性极高，属于高频重度用户。

### 2. 返回分析（回访间隔）
- **分析逻辑**：计算用户相邻两次活跃（去重后的日期）之间的天数差。
- **数据发现**：隔 1 天回访的用户高达 9,838 人，随后随天数增加迅速衰减。
- **业务洞察**：用户具有典型的“日活跃”特征。隔天回访人数极高，说明用户习惯每天打开平台浏览。这也印证了前面漏斗分析中，`pv -> pv` 流量极大的原因。
