## 🚀 项目阶段一：用户行为转化漏斗分析

### 1. 业务背景
本次分析基于淘宝用户行为数据集（采样数据），旨在洞察用户从“浏览商品”到“最终购买”的转化路径，寻找流失最严重的环节，为运营和产品优化提供数据支持。

### 2. 数据说明
- 数据表：`user_behavior_clean`（清洗后数据，剔除了 2017-11-25 至 2017-12-03 之外的异常记录）。
- 统计口径：**独立用户数（UV）**，而非行为次数（PV），避免“一人点击百次”导致数据失真。

### 3. 核心发现（漏斗各环节数据）
| 行为类型 | 独立用户数 (UV) | 相对 PV 转化率 |
| :--- | :--- | :--- |
| 浏览 (pv) | 94 | 100.00% |
| 加购 (cart) | 73 | 77.66% |
| 收藏 (fav) | 37 | 39.36% |
| 购买 (buy) | 60 | 63.83% |

*(注：当前为 LIMIT 10000 测试样本，数字较小，后续将跑全量数据复测)*

### 4. 业务洞察
- **加购转化率极高（77.66%）**：说明“浏览 -> 加购”的页面引导做得极好，用户把商品放入购物车的意愿强烈。
- **购买转化率可观（63.83%）**：说明从加购到结算的流程体验较顺畅，用户的下单决策较果断。
- **收藏行为较低（39.36%）**：收藏作为“犹豫期”的标记行为，其人数少于加购，说明大部分用户偏向直接加购，而非对比收藏。

### 5. 下一步计划
- 拉取全量数据重新验证漏斗比例。
- 使用 `LAG()` 做用户路径分析，看看“收藏未购买”的用户最终流向了哪里。
